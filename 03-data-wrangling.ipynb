# 03 - Data Wrangling
Explore the dataset, count launches per site/orbit, and create the binary landing `Class` label.

```
import pandas as pd, numpy as np
df=pd.read_csv('https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DS0701EN-SkillsNetwork/api/dataset_part_1.csv')
df.head()
```

```
print(df.isnull().sum()/len(df)*100)
print(df.dtypes)
```

```
print(df['LaunchSite'].value_counts())
print(df['Orbit'].value_counts())
landing_outcomes=df['Outcome'].value_counts()
print(landing_outcomes)
```

```
bad_outcomes=set(landing_outcomes.keys()[[1,3,5,6,7]])
print(bad_outcomes)
df['Class']=[0 if o in bad_outcomes else 1 for o in df['Outcome']]
print('Success rate:',df['Class'].mean())
df.to_csv('dataset_part_2.csv',index=False)
```
