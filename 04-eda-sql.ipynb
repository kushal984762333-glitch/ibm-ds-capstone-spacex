# 04 - EDA with SQL
Loads the SpaceX table into SQLite and answers the lab questions.

```
import sqlite3, pandas as pd
con=sqlite3.connect('my_data1.db')
df=pd.read_csv('https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/labs/module_2/data/Spacex.csv')
df.to_sql('SPACEXTBL',con,if_exists='replace',index=False)
def q(sql): return pd.read_sql(sql,con)
q('select * from SPACEXTBL limit 3')
```

```
print(q('select distinct Launch_Site from SPACEXTBL'))
print(q("select * from SPACEXTBL where Launch_Site like 'CCA%' limit 5"))
```

```
print(q("select sum(PAYLOAD_MASS__KG_) total_nasa_crs from SPACEXTBL where Customer='NASA (CRS)'"))
print(q("select avg(PAYLOAD_MASS__KG_) avg_f9v11 from SPACEXTBL where Booster_Version like 'F9 v1.1%'"))
print(q("select min(Date) first_ground_landing from SPACEXTBL where Landing_Outcome like '%ground pad%'"))
```

```
print(q("select distinct Booster_Version from SPACEXTBL where Landing_Outcome='Success (drone ship)' and PAYLOAD_MASS__KG_ between 4000 and 6000"))
print(q('select Mission_Outcome,count(*) n from SPACEXTBL group by Mission_Outcome'))
print(q('select distinct Booster_Version from SPACEXTBL where PAYLOAD_MASS__KG_=(select max(PAYLOAD_MASS__KG_) from SPACEXTBL)'))
```

```
print(q("select substr(Date,6,2) month,Landing_Outcome,Booster_Version,Launch_Site from SPACEXTBL where Landing_Outcome='Failure (drone ship)' and substr(Date,0,5)='2015'"))
print(q("select Landing_Outcome,count(*) n from SPACEXTBL where Date between '2010-06-04' and '2017-03-20' group by Landing_Outcome order by n desc"))
```
