# 06 - Interactive Map with Folium
Markers for launch sites, success/failure clusters, and proximity analysis.

```
import folium, pandas as pd
from folium.plugins import MarkerCluster, MousePosition
from folium.features import DivIcon
from math import sin,cos,sqrt,atan2,radians
spacex_df=pd.read_csv('https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/spacex_launch_geo.csv')
sites=spacex_df.groupby('Launch Site',as_index=False).first()[['Launch Site','Lat','Long']]
m=folium.Map(location=[29.559684888503615,-95.0830971930759],zoom_start=5)
for _,r in sites.iterrows():
    loc=[r['Lat'],r['Long']]
    folium.Circle(loc,radius=1000,color='#d35400',fill=True).add_child(folium.Popup(r['Launch Site'])).add_to(m)
    folium.map.Marker(loc,icon=DivIcon(icon_size=(20,20),icon_anchor=(0,0),html='<div style="font-size:12;color:#d35400"><b>%s</b></div>'%r['Launch Site'])).add_to(m)
m
```

```
spacex_df['marker_color']=spacex_df['class'].apply(lambda c:'green' if c==1 else 'red')
cluster=MarkerCluster();m.add_child(cluster)
for _,r in spacex_df.iterrows():
    cluster.add_child(folium.Marker([r['Lat'],r['Long']],icon=folium.Icon(color='white',icon_color=r['marker_color'])))
m
```

```
def dist(lat1,lon1,lat2,lon2):
    R=6373.0;a=sin(radians(lat2-lat1)/2)**2+cos(radians(lat1))*cos(radians(lat2))*sin(radians(lon2-lon1)/2)**2
    return R*2*atan2(sqrt(a),sqrt(1-a))
site=(28.563197,-80.576820)  # CCAFS SLC-40
coast=(28.56367,-80.57163)   # adjust using MousePosition on your map
d=dist(*site,*coast);print('distance to coast (km):',round(d,2))
folium.PolyLine([site,coast],weight=1).add_to(m)
folium.Marker(coast,icon=DivIcon(icon_size=(20,20),icon_anchor=(0,0),html='<div style="font-size:12;color:#252526"><b>%.2f KM</b></div>'%d)).add_to(m)
m
# Repeat for nearest railway, highway and city; save screenshots to images/
```
